```python
#import Pandas Library
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Assigned dataset variable to read insurance.csv file with the function of read_csv
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
#Convert Categorical data to nominal data (one hot encoding) and drop first categorical (State_California)
dataset=pd.get_dummies(dataset,drop_first=True).astype(int)
```

```python
dataset
```

```python
#Find column Headers
dataset.columns
```

```python
#Assign variable (independent) for only input data
independent=dataset[['age', 'bmi', 'children', 'charges', 'sex_male', 'smoker_yes']]
```

```python
independent
```

```python
#Assign variable for Output (dependent)
dependent=dataset[['charges']]
```

```python
dependent
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
param_grid = {"criterion":["squared_error","absolute_error", "friedman_mse", "poisson"],
    "max_features":[None, "sqrt", "log2"],
    "splitter":["best", "random"]
}
grid = GridSearchCV(DecisionTreeRegressor(random_state=0), param_grid, cv=5, refit=True, verbose=3, n_jobs=-1, error_score="raise")
#fitting the model for grid search
grid.fit(independent, dependent)
```

```python
#print best parameter after tuning
#print(grid.best_params_)
re=grid.cv_results_
print ("The R_Score value for best parameter {}:".format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
age_input=float(input("Age:"))
bmi_input=float(input("BMI:"))
children_input=float(input("Children:"))
sex_male_input=int(input("Sex Male <0 or 1>:"))
sex_female_input=int(input("Sex Female <0 or 1>:"))
smoker_yes_input=int(input("Smoker Yes <0 or 1>:")) 
```

```python
Future_Prediction = grid.predict([[age_input, bmi_input, children_input, sex_male_input, smoker_yes_input, sex_female_input]])
print("Future_Prediction={}".format(Future_Prediction))
print("Best Parameters:", grid.best_params_)
print("Best Score:", grid.best_score_)
```
